```python
import numpy as np 
import pandas as pd

data = pd.read_csv("sales_data_sample.csv",encoding="latin1")

arr = data.to_numpy()
```

```python
print("Datatype: ", arr.dtype)

print("SHAPE:",arr.shape)
print("ROWS:",arr.shape[0])
print("COLUMNS:",arr.shape[1])

print("First Five Rows:\n",arr[0:5])
print("Last Five Rows:\n",arr[-5:])
print("First Five columns:\n", arr[:, 0:5])
```

```python
print("10th ROW: \n", arr[9])
print("10th Column: \n", arr[:,9])
```

```python
print("Qunatity+sales :\n", arr[:, [1,4]])

quan_ord = arr[:,1].astype(float)
print(arr[quan_ord > 50])
sales_ord = arr[:,4].astype(float)
print(arr[sales_ord>10000])
```

```python
condition = (quan_ord > 30) & (sales_ord > 5000)
print(arr[condition])
```

```python
# Task 11

# Calculate the:
# Total sales
# Average sales
# Minimum sales
# Maximum sales

print("Total Sales: ", np.sum(sales_ord))
print("Average Sales: ", np.mean(sales_ord))
print("Minimum Sales: ", np.min(sales_ord))
print("Maximum Sales: ", np.max(sales_ord))
```

```python
# Task 12

# Calculate the:

# Total quantity ordered
# Average quantity ordered
# Minimum quantity
# Maximum quantity

print("Total Quantity Order: ", np.sum(quan_ord))
print("Average Quantity Order: ", np.mean(quan_ord))
print("Minimum Quantity Order: ", np.min(quan_ord))
print("Maximum Quantity Order: ", np.max(quan_ord))
```

```python
# Task 13

# Find the standard deviation of Sales.

print("STD: ", np.std(sales_ord))

# Task 14

# Find the median Sales.
print("Median: ", np.median(sales_ord))
# Task 15

# Calculate:
# Total Revenue = sum of all Sales
# Then verify it using another NumPy approach.

total_revenue = np.sum(sales_ord)
print("Total Revenue : ", total_revenue)

sum_of_all_sales = np.mean(sales_ord) * np.size(sales_ord)
print("sum of all sales = :", sum_of_all_sales)
```

```python
# Task 16
# Count how many orders have:
# SALES > 5000

print("SALES > 5000:", len(arr[sales_ord > 5000]))
print("Sales > 5000: ", np.sum(sales_ord > 5000))

# Task 17
# Count how many orders have:
# QUANTITYORDERED > 40

print("QUANTITY ORDERED > 40 : ", len(arr[quan_ord > 40]))
print("Quantity ordered > 5000: ", np.sum(quan_ord > 40))

# Task 18
# Find how many orders satisfy:
# SALES > 5000 AND QUANTITYORDERED > 40

condition = (sales_ord > 5000) & (quan_ord > 40)
print("SALES > 5000 AND QUANTITYORDERED > 40: ", len(arr[condition]))
print("SALES > 5000 AND QUANTITYORDERED > 40: ", np.sum(condition))

 
```

```python
# Task 19
# Find the percentage of orders where:
# SALES > 5000

order_sales = arr[:, 4].astype(float)
order_sales_5000 = order_sales[order_sales > 5000]
percentage = (np.sum(order_sales > 5000) * 100) / np.size(order_sales)
print("Percentage of orders where sales > 5000: ", percentage)
```

```python
# 🔴 Level 5 — Challenge
# Try these without looking up the solution.
# Task 20
# Find the row containing the highest Sales.
# Then extract the complete row.

sales_ord = arr[:, 4].astype(float)
row_highest_sale = np.argmax(sales_ord)
print("Row containing highest sales : ", row_highest_sale)
print("Complete Row containing highest sales : ", arr[row_highest_sale])


# Task 21
# Find the row containing the lowest Sales.
row_lowest_sale = np.argmin(sales_ord)
print("Row containing lowest sales : ", row_lowest_sale)
print("Complete Row containing lowest sales : ", arr[row_lowest_sale])

# Task 22
# Find the index/position of the order with the highest Sales.
print("Index containing highest sales : ", row_highest_sale)

# Task 23
# Find the top 10 highest sales values using NumPy.
sort_sales = np.sort(sales_ord)
print("top 10 highest sales :", sort_sales[-10:])

# Task 24
# Calculate the difference between:
# Maximum Sales − Minimum Sales
diff = np.max(sales_ord) - np.min(sales_ord)
print("Maximum Sales − Minimum Sales:", diff)

```

```python
# Task 25 ⭐
# Create a new NumPy array containing:
# QUANTITYORDERED | PRICEEACH | SALES
# Then calculate the correlation between:
# Quantity Ordered and Sales
# Price Each and Sales
quan_arr = arr[:,1].astype(float)
priceeach_arr = arr[:,2].astype(float)
sales_arr = arr[:,4].astype(float)

# new_arr = np.concatenate((quan_arr, priceeach_arr, sales_arr), axis=1) # Not working

new_arr = np.column_stack((quan_arr, priceeach_arr, sales_arr))

correlation1 = np.corrcoef(quan_arr, sales_arr)
correlation2 = np.corrcoef(priceeach_arr, sales_arr)
print(new_arr)
print("Correlation between Quantity Ordered and Sales:", correlation1)
print("Correlation between Quantity Ordered and Sales:", correlation1[0,1])
print("Correlation between price each and Sales:", correlation2)
print("Correlation between price each and Sales:", correlation2[0,1])
```
