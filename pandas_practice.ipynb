```python
df.columns.tolist()
df["ADDRESSLINE2"].isna().sum()
df["ADDRESSLINE2"].isna().sum()
df["TERRITORY"].isna().sum()
```

```python
df2 = df.copy()

df2 = df2.drop(columns=["TERRITORY"])

#OR 

df2.drop(columns=["ADDRESSLINE2"], inplace=True)
```

```python
df2.columns.tolist()
```

```python
# Filling missing values

# df2.info()
df2["STATE"].isna().sum()

# df2 = df2["STATE"].fillna("")
# examplle
# df2["AGE"] = df2["AGE"].fillna(df2["AGE"].mean())  # fill by mean
# df2["AGE"] = df2["AGE"].fillna(df2["AGE"].median())  # fill by median
# df2["AGE"] = df2["AGE"].fillna(df2["AGE"].mode())  # fill by mode
```

```python
# AS we know state is a categorical column so replace nan with the mode makes sense

df2["STATE"] = df2["STATE"].fillna(df["STATE"].mode()[0])

df2["STATE"].isna().sum()
```

```python
df2.dtypes


```

```python
# df2["POSTALCODE"].astype(int) # not working because NAn to integer not working so we use below

df2["POSTALCODE"] = pd.to_numeric(df["POSTALCODE"], errors="coerce")

df2["POSTALCODE"].isna().sum()

df2["POSTALCODE"] = df2["POSTALCODE"].ffill() # Forward Fill
```

```python
df2.dtypes
df2["POSTALCODE"].isna().sum()
```

```python
#ORDER DATE is of object type we have to convert it into date

df2["ORDERDATE"]

df2["ORDERDATE"] = pd.to_datetime(df["ORDERDATE"])

df2["ORDERDATE"].dt.year
df2["ORDERDATE"].dt.month
df2["ORDERDATE"].dt.date
df2["ORDERDATE"].dt.day
```

```python
# DATA MANIPULATION

df2["TOTAL_PRICE"] = df2["QUANTITYORDERED"] * df2["PRICEEACH"]
```

```python
# df2 = df2.drop(columns=["TOTAL_PRICE"])

# df2 = df2.rename(columns={"SALES": "PRICE"})

```

```python
#Replace

df2["DEALSIZE"] = df2["DEALSIZE"].replace({
    "Small":"S",
    "Medium":"M",
    "Large":"L"
})
```

```python
#Map

df2["DEALSIZE"] = df2["DEALSIZE"].map({
    "S":"Small",
    "M":"Medium",
    "L":"Large"
})
```

```python
# Sorting 

df2 =  df2.sort_values("SALES") # Ascending
# df2 = df2.sort_values("SALES", ascending=False) # descending
```

```python
# GroupBy and Aggregation

#Average Sales for each country
df2.groupby("COUNTRY")["SALES"].mean() 

#Total Sales for each country
df2.groupby("COUNTRY")["SALES"].sum()
df2.groupby("COUNTRY")["SALES"].min()
df2.groupby("COUNTRY")["SALES"].max()
df2.groupby("COUNTRY")["SALES"].count()


#Multiple Aggregation

df2.groupby("COUNTRY")["SALES"].agg(["mean", "max", "min", "sum"])

df2.groupby("COUNTRY").agg(
    avg_sales=("SALES", "mean"),
    sum_sales=("SALES", "sum"),
    min_sales=("SALES", "min"),
    max_sales=("SALES", "max")
)

df2.groupby("COUNTRY", as_index= False)["SALES"].mean()
```
